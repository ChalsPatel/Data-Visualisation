```python
# ============================================================
# IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import plotly.express as px
import folium

from folium.plugins import HeatMap, MarkerCluster
from wordcloud import WordCloud
```

```python



# ============================================================
# QUESTION 1
# Total deaths per year from 1984 to 2016
# ============================================================

# Filter the required years
df_1984_2016 = df[
    (df["year"] >= 1984) &
    (df["year"] <= 2016)
]

# Count deaths per year
yearly_deaths = (
    df_1984_2016
    .groupby("year")
    .size()
    .reset_index(name="total_deaths")
)

# Create bar chart
fig = px.bar(
    yearly_deaths,
    x="year",
    y="total_deaths",
    title="Total Police Deaths per Year (1984–2016)",
    labels={
        "year": "Year",
        "total_deaths": "Total Deaths"
    }
)

fig.update_layout(
    xaxis=dict(dtick=1),
    template="plotly_white"
)

fig.show()
```

```python
# ============================================================
# QUESTION 2
# Line chart comparing yearly deaths in different states
# ============================================================

# Remove extra spaces from state names
df["state"] = df["state"].str.strip()

# Filter 1984–2016
df_1984_2016 = df[
    (df["year"] >= 1984) &
    (df["year"] <= 2016)
]

# Find top 10 states by total deaths
top_states = (
    df_1984_2016
    .groupby("state")
    .size()
    .sort_values(ascending=False)
    .head(10)
    .index
)

# Keep only top 10 states
state_yearly = (
    df_1984_2016[
        df_1984_2016["state"].isin(top_states)
    ]
    .groupby(["year", "state"])
    .size()
    .reset_index(name="deaths")
)

# Create line chart
fig = px.line(
    state_yearly,
    x="year",
    y="deaths",
    color="state",
    markers=True,
    title="Yearly Police Deaths by State",
    labels={
        "year": "Year",
        "deaths": "Number of Deaths",
        "state": "State"
    }
)

fig.update_layout(
    template="plotly_white",
    xaxis=dict(dtick=2)
)

fig.show()
```

```python
# ============================================================
# QUESTION 7
# Deaths by cause, animated by year
# ============================================================

# Filter years
cause_year = df[
    (df["year"] >= 1984) &
    (df["year"] <= 2016)
].copy()

# Count deaths by year and cause
cause_year = (
    cause_year
    .groupby(["year", "cause_short"])
    .size()
    .reset_index(name="deaths")
)

# Keep major causes
top_causes = (
    cause_year
    .groupby("cause_short")["deaths"]
    .sum()
    .sort_values(ascending=False)
    .head(15)
    .index
)

cause_year = cause_year[
    cause_year["cause_short"].isin(top_causes)
]

# Animated bar chart
fig = px.bar(
    cause_year,
    x="deaths",
    y="cause_short",
    color="cause_short",
    animation_frame="year",
    orientation="h",
    title="Police Deaths by Cause — Animated by Year",
    labels={
        "deaths": "Number of Deaths",
        "cause_short": "Cause of Death"
    }
)

fig.update_layout(
    showlegend=False,
    template="plotly_white"
)

fig.show()
```

```python
# ============================================================
# QUESTION 8
# Heatmap of deaths by state and year
# ============================================================

# Filter 1984–2016
state_year = df[
    (df["year"] >= 1984) &
    (df["year"] <= 2016)
].copy()

# Create state-year table
state_year_table = pd.crosstab(
    state_year["state"],
    state_year["year"]
)

# Plot heatmap
plt.figure(figsize=(20, 18))

sns.heatmap(
    state_year_table,
    cmap="YlOrRd",
    linewidths=0.2,
    linecolor="white"
)

plt.title(
    "Police Deaths by State and Year (1984–2016)",
    fontsize=18
)

plt.xlabel("Year")
plt.ylabel("State")

plt.tight_layout()

plt.show()
```

```python
# ============================================================
# QUESTION 9
# Treemap: State → Department → Cause
# ============================================================

# Remove spaces from state
treemap_df = df.copy()

treemap_df["state"] = (
    treemap_df["state"]
    .astype(str)
    .str.strip()
)

# Remove unnecessary missing values
treemap_df = treemap_df.dropna(
    subset=["state", "dept_name", "cause_short"]
)

# Count deaths
treemap_data = (
    treemap_df
    .groupby(
        ["state", "dept_name", "cause_short"]
    )
    .size()
    .reset_index(name="deaths")
)

# Create treemap
fig = px.treemap(
    treemap_data,
    path=[
        "state",
        "dept_name",
        "cause_short"
    ],
    values="deaths",
    title="Police Deaths: State → Department → Cause",
    color="deaths",
    color_continuous_scale="Reds"
)

fig.show()
```
